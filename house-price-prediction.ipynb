```python
Train=pd.read_csv("/kaggle/input/competitions/home-data-for-ml-course/train.csv")
Train.describe()
```

```python
Test=pd.read_csv("/kaggle/input/competitions/home-data-for-ml-course/test.csv")
Test.head()
```

```python
#choosing target and feature
y=Train.SalePrice
feature=["LotArea","MoSold","YrSold","MSSubClass","Id"]
X=Train[feature]
X_Test=Test[feature]

#import metrics,train_test-split,randomforestregressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_absolute_error
from sklearn.model_selection import  train_test_split

#splitting the target and feature
train_X,val_X,train_y,val_y=train_test_split(X,y,random_state=0)

#defining the model
model_test=RandomForestRegressor(random_state=0)

#fitting the model
model_test.fit(train_X,train_y)

#predicting 
prediction=model_test.predict(val_X)

#evaluating the model
mae=mean_absolute_error(val_y,prediction)

print(Train.SalePrice.mean())
print(mae)

#final model
model=RandomForestRegressor(random_state=0)

#final fit
model.fit(X,y)

#predicting
predictions=model.predict(X_Test)
print(predictions)

#trying decisiontree
max_mae=None
best_leaf=None

#trying different depth in a loop
for depth in [5,15,25,100,250,500]:

    #define decision model
    model_tree=DecisionTreeRegressor(max_depth=depth,random_state=0)
    #fitting the decisionmodel
    model_tree.fit(train_X,train_y)
    #predicting model
    prediction_tree=model_tree.predict(val_X)
    mae_tree=mean_absolute_error(val_y,prediction_tree)

    #checking the best depth
    if max_mae is None or mae_tree<max_mae:
        max_mae=mae_tree
        best_leaf=depth
        print(f"max_mae is {max_mae} and best leaf is {depth}")

#final model-tree
final_model_tree=DecisionTreeRegressor(max_depth=5,random_state=0)

#fitting model_tree
final_model_tree.fit(X,y)

#prediction final model tree
final_prediction_tree=final_model_tree.predict(X_Test)
print(final_prediction_tree)

        
                                                                                             

    
```
